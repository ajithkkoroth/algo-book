```python
import ipywidgets as widgets
from IPython.display import display
import base64
import math

# Helper to decode Base64 strings
def b64_dec(s):
    return base64.b64decode(s).decode('utf-8')

# ==========================================================
# 1. QUESTION BANK: STAGE 1 (MCQs)
# ==========================================================
mcq_data = [
    {
        "title": "1. Problem-Solving Strategies",
        "prompt": "You lose your hostel keys. Instead of searching every inch of campus, you check the canteen table and your pocket first. Which strategy is this?",
        "code": "",
        "options": ["Trial and Error", "Heuristics (Rule of Thumb)", "Working Backward", "Problem Decomposition"],
        "b64_correct": "SGV1cmlzdGljcyAoUnVsZSBvZiBUaHVtYik=",
        "feedback": {
            "Trial and Error": "Not quite! Trial and Error checks every possible spot without an educated guess.",
            "Heuristics (Rule of Thumb)": "Correct! Using past experience to check the most likely places first is a Heuristic.",
            "Working Backward": "Not quite! Working backward starts from the end goal (like an exam time) and subtracts.",
            "Problem Decomposition": "Not quite! Decomposition breaks a massive problem into smaller sub-problems."
        }
    },
    {
        "title": "2. The Text-to-Math Trap",
        "prompt": "What gets printed when Python runs these two lines?",
        "code": 'price = "45"\nprint(price + "5")',
        "options": ["50", "455", "price + 5", "TypeError"],
        "b64_correct": "NDU1",
        "feedback": {
            "50": "Careful! Look at the quotes around '45' and '5'. Python treats them as Strings (text), not numbers.",
            "455": "Spot on! Because both are Strings inside quotes, the + operator glues them side-by-side into '455'.",
            "price + 5": "price is not inside quotes in print(), so Python looks inside the price box.",
            "TypeError": "Gluing two Strings together with + is completely legal in Python."
        }
    },
    {
        "title": "3. Floor Division vs. Modulo",
        "prompt": "You divide 17 samosas equally among 5 friends. Which expression gives the LEFT OVER samosas?",
        "code": "",
        "options": ["17 / 5", "17 // 5", "17 % 5", "17 ** 5"],
        "b64_correct": "MTcgJSA1",
        "feedback": {
            "17 / 5": "17 / 5 gives 3.4 (decimal division).",
            "17 // 5": "Floor division (//) gives 3 (how many each friend eats, not what is left over).",
            "17 % 5": "Correct! Modulo (%) gives the remainder of division: 17 % 5 = 2 leftover samosas.",
            "17 ** 5": "** calculates 17 to the power of 5!"
        }
    },
    {
        "title": "4. Operator Precedence (BODMAS)",
        "prompt": "Act as the computer. What exact number is stored in result?",
        "code": "result = 10 - 2 * 3 ** 2",
        "options": ["64", "72", "-8", "-26"],
        "b64_correct": "LTg=",
        "feedback": {
            "64": "Remember BODMAS! Do not subtract 10 - 2 first.",
            "72": "Python calculates 3 ** 2 (9) first, then 2 * 9 (18), then 10 - 18.",
            "-8": "Correct! Step 1: 3 ** 2 = 9. Step 2: 2 * 9 = 18. Step 3: 10 - 18 = -8.",
            "-26": "2 * 3 ** 2 squares only the 3, not (2 * 3)."
        }
    },
    {
        "title": "5. The Logic Gates",
        "prompt": "Trace the bouncers! What does the final line print?",
        "code": "has_id = True\npaid_fee = False\nprint(has_id and not paid_fee)",
        "options": ["True", "False", "None", "SyntaxError"],
        "b64_correct": "VHJ1ZQ==",
        "feedback": {
            "True": "Correct! Since paid_fee is False, 'not paid_fee' flips it to True. Then 'True and True' is True!",
            "False": "Look closely at the 'not' operator! paid_fee is False, so 'not paid_fee' flips to True.",
            "None": "Logical operators always evaluate to True or False, never None.",
            "SyntaxError": "Combining 'and' with 'not' is completely valid Python syntax."
        }
    }
]

# ==========================================================
# 2. QUESTION BANK: STAGE 2 (PSEUDOCODE & FLOWCHARTS)
# ==========================================================
pseudo_tasks = [
    {
        "title": "1. Area of a Circle (AI Checkpoint 1.1)",
        "prompt": "Write a 4-step pseudocode algorithm to READ radius, COMPUTE area = 3.14159 * radius * radius, and PRINT area. (Include START and STOP!)",
        "b64_model": "U1RBUlQKUkVBRCBSYWRpdXMKQ09NUFVURSBhcmVhID0gMy4xNDE1OSAqIFJhZGl1cyAqIFJhZGl1cwpQUklOVCBhcmVhClNUT1A=",
        "flow": [("oval", "START"), ("io", "READ radius"), ("proc", "COMPUTE area = 3.14159 * radius * radius"), ("io", "PRINT area"), ("oval", "STOP")]
    },
    {
        "title": "2. Weeks & Leftover Days (Level 3 Mission)",
        "prompt": "Write pseudocode to READ total_days, compute full weeks using // 7, compute leftover days using % 7, and PRINT both.",
        "b64_model": "U1RBUlQKUkVBRCB0b3RhbF9kYXlzCkNPTVBVVEUgd2Vla3MgPSB0b3RhbF9kYXlzIC8vIDcKQ09NUFVURSBsZWZ0b3ZlciA9IHRvdGFsX2RheXMgJSA3ClBSSU5UIHdlZWtzLCBsZWZ0b3ZlcgpTVE9Q",
        "flow": [("oval", "START"), ("io", "READ total_days"), ("proc", "COMPUTE weeks = total_days // 7"), ("proc", "COMPUTE leftover = total_days % 7"), ("io", "PRINT weeks, leftover"), ("oval", "STOP")]
    },
    {
        "title": "3. Hostel Currency Converter (INR to USD)",
        "prompt": "Write pseudocode to READ amount_inr, COMPUTE amount_usd = amount_inr / 83, and PRINT amount_usd.",
        "b64_model": "U1RBUlQKUkVBRCBhbW91bnRfaW5yCkNPTVBVVEUgYW1vdW50X3VzZCA9IGFtb3VudF9pbnIgLyA4MwpQUklOVCBhbW91bnRfdXNkClNUT1A=",
        "flow": [("oval", "START"), ("io", "READ amount_inr"), ("proc", "COMPUTE amount_usd = amount_inr / 83"), ("io", "PRINT amount_usd"), ("oval", "STOP")]
    }
]

# ==========================================================
# 3. QUESTION BANK: STAGE 3 (PYTHON MISSIONS & TEST TABLES)
# ==========================================================
def build_test_table(rows):
    tr_html = ""
    for tc_num, test_in, exp_out, act_out, passed, comment in rows:
        badge = "<span style='color:#166534; font-weight:bold;'>&#9989; PASS</span>" if passed else "<span style='color:#991b1b; font-weight:bold;'>&#10060; FAIL</span>"
        tr_html += f"""
        <tr>
          <td>#{tc_num}</td>
          <td><code>{test_in}</code></td>
          <td><code>{exp_out}</code></td>
          <td><code>{act_out}</code></td>
          <td>{badge}</td>
          <td style='font-family:sans-serif;'>{comment}</td>
        </tr>
        """
    return f"""
    <div class='table-scroll'>
      <table class='test-table'>
        <thead>
          <tr>
            <th>#</th>
            <th>Input</th>
            <th>Expected</th>
            <th>Yours</th>
            <th>Status</th>
            <th>Feedback</th>
          </tr>
        </thead>
        <tbody>{tr_html}</tbody>
      </table>
    </div>
    """

def grade_mission_1(code):
    test_cases = [
        (1, '100', '105', 105, 'Standard integer input'),
        (2, '45',  '50',  50,  'Odd number price check'),
        (3, '0',   '5',   5,   'Zero-rupee edge case')
    ]
    rows = []
    passed_count = 0
    for num, tin, exp_str, exp_val, desc in test_cases:
        ns = {"input": lambda p="": tin}
        out = []
        ns["print"] = lambda *a: out.append(" ".join(str(x) for x in a))
        try:
            exec(code, ns)
            actual = out[-1].strip() if out else "Nothing"
            passed = (actual == exp_str and ns.get("final_bill") == exp_val)
            comment = desc if passed else "Use int(input())"
        except TypeError:
            actual = "TypeError"
            passed = False
            comment = "Wrap input() in int()"
        except Exception as e:
            actual = f"{type(e).__name__}"
            passed = False
            comment = str(e)
        if passed:
            passed_count += 1
        rows.append((num, tin, exp_str, actual, passed, comment))
    return (passed_count == 3), passed_count, build_test_table(rows)

def grade_mission_2(code):
    test_cases = [
        (1, '10', '1 3', 1, 3, '10 days = 1w, 3d'),
        (2, '28', '4 0', 4, 0, 'Exact multiple of 7'),
        (3, '5',  '0 5', 0, 5, 'Less than 1 week')
    ]
    rows = []
    passed_count = 0
    for num, tin, exp_str, ew, el, desc in test_cases:
        ns = {"input": lambda p="": tin}
        out = []
        ns["print"] = lambda *a: out.append(" ".join(str(x) for x in a))
        try:
            exec(code, ns)
            actual = out[-1].strip() if out else "Nothing"
            if isinstance(ns.get("weeks"), float):
                passed = False
                comment = "Use // instead of /"
            else:
                passed = (ns.get("weeks") == ew and ns.get("leftover") == el and actual == exp_str)
                comment = desc if passed else f"Exp: {ew} and {el}"
        except Exception as e:
            actual = f"{type(e).__name__}"
            passed = False
            comment = str(e)
        if passed:
            passed_count += 1
        rows.append((num, tin, exp_str, actual, passed, comment))
    return (passed_count == 3), passed_count, build_test_table(rows)

def grade_mission_3(code):
    test_cases = [
        (1, '5',   '78.5398', 78.53981633974483, 'Integer radius (r=5)'),
        (2, '2.5', '19.6350', 19.634954084936208, 'Float radius (r=2.5)'),
        (3, '0',   '0.0000',  0.0,                'Zero radius (r=0)')
    ]
    rows = []
    passed_count = 0
    for num, tin, exp_disp, exp_val, desc in test_cases:
        ns = {"input": lambda p="": tin}
        out = []
        ns["print"] = lambda *a: out.append(" ".join(str(x) for x in a))
        if "import math" not in code or "math.pi" not in code:
            rows.append((num, tin, exp_disp, "No math.pi", False, "Use import math & math.pi"))
            continue
        try:
            exec(code, ns)
            area_val = ns.get("area")
            actual = f"{area_val:.4f}" if isinstance(area_val, (int, float)) else "Missing"
            passed = isinstance(area_val, (int, float)) and abs(area_val - exp_val) < 0.001
            comment = desc if passed else "Use math.pi * (radius ** 2)"
        except Exception as e:
            actual = f"{type(e).__name__}"
            passed = False
            comment = str(e)
        if passed:
            passed_count += 1
        rows.append((num, tin, exp_disp, actual, passed, comment))
    return (passed_count == 3), passed_count, build_test_table(rows)

missions_spec = [
    (
        "Mission 1: Fix the Text-to-Math Trap",
        "Wrap <code>input()</code> inside <code>int()</code> on line 1 so all 3 test cases pass.",
        'price = input("Enter item price: ")\ntax = 5\nfinal_bill = price + tax\nprint(final_bill)',
        grade_mission_1,
        "Change line 1 to: <code>price = int(input(\"Enter item price: \"))</code>"
    ),
    (
        "Mission 2: The Week & Day Splitter",
        "Calculate full 7-day <code>weeks</code> using <code>// 7</code> and <code>leftover</code> days using <code>% 7</code>.",
        'days = int(input("Enter days: "))\nweeks = \nleftover = \nprint(weeks, leftover)',
        grade_mission_2,
        "Set <code>weeks = days // 7</code> and <code>leftover = days % 7</code>."
    ),
    (
        "Mission 3: Exact Circle Area Calculator",
        "Import <code>math</code> at the top and calculate <code>area = math.pi * (radius ** 2)</code>.",
        '# Step 1: Import math\n\nradius = float(input("Enter radius: "))\narea = \nprint(area)',
        grade_mission_3,
        "Write <code>import math</code> on line 1, and <code>area = math.pi * (radius ** 2)</code>."
    )
]

# ==========================================================
# AUTO-SCALING PROGRESS ENGINE (NATIVE VBOX = ZERO CROPPING!)
# ==========================================================
NUM_MCQ = len(mcq_data)
NUM_PSEUDO = len(pseudo_tasks)
NUM_PY = len(missions_spec)

state = {
    "mcq_passed": [False] * NUM_MCQ,
    "pseudo_done": [False] * NUM_PSEUDO,
    "py_passed": [False] * NUM_PY,
    "py_tc_scores": [0] * NUM_PY,
    "py_attempts": [0] * NUM_PY
}

header_widget = widgets.HTML(value="""
<style>
  /* Minimal Compact Header */
  .lab-header { background: linear-gradient(135deg, #1e3a8a, #1e40af); color: white; padding: 12px 14px; border-radius: 8px; margin-bottom: 8px; font-family: sans-serif; }
  .book-badge { display: inline-block; background: rgba(255,255,255,0.18); color: #e0f2fe; padding: 2px 8px; border-radius: 12px; font-size: 0.7rem; font-weight: 700; text-transform: uppercase; margin-bottom: 4px; }
  .lab-header h1 { margin: 0 0 2px 0; font-size: 1.15rem; color: white; line-height: 1.2; }
  .author-line { margin: 0; font-size: 0.8rem; color: #bfdbfe; font-weight: 600; }

  /* Shrink Tab Widths & Padding to Minimal Size */
  .lm-TabBar-tab, .p-TabBar-tab, .jupyter-widgets.widget-tab > .p-TabBar .p-TabBar-tab {
    min-width: 65px !important;
    max-width: 115px !important;
    padding: 4px 8px !important;
    font-size: 0.78rem !important;
    font-family: sans-serif !important;
  }

  /* Single-line compact row inside native VBox */
  .prog-line {
    font-family: sans-serif;
    font-size: 0.82rem;
    font-weight: 600;
    color: #1e293b;
    line-height: 1.3;
    padding: 1px 4px;
  }
  .prog-line-score {
    font-family: sans-serif;
    font-size: 0.84rem;
    font-weight: 700;
    color: #1e40af;
    line-height: 1.3;
    padding: 2px 4px 0 4px;
    border-top: 1px solid #cbd5e1;
  }

  .code-card { background: #1e293b; color: #f8fafc; padding: 10px; border-radius: 6px; font-family: monospace; font-size: 0.85rem; white-space: pre-wrap; word-break: break-word; margin: 6px 0; line-height: 1.35; }
  .pass-box { background: #dcfce7; color: #166534; padding: 8px 10px; border-radius: 6px; font-weight: 600; font-size: 0.85rem; margin-top: 6px; border: 1px solid #86efac; font-family: sans-serif; }
  .fail-box { background: #fee2e2; color: #991b1b; padding: 8px 10px; border-radius: 6px; font-weight: 600; font-size: 0.85rem; margin-top: 6px; border: 1px solid #fca5a5; font-family: sans-serif; }
  .hint-box { background: #fef9c3; color: #854d0e; padding: 8px 10px; border-radius: 6px; font-weight: 500; font-size: 0.85rem; margin-top: 6px; border: 1px solid #fde047; font-family: sans-serif; }
  .lock-banner { background: #f1f5f9; color: #475569; padding: 16px 12px; border-radius: 8px; border: 2px dashed #94a3b8; text-align: center; font-family: sans-serif; font-size: 0.9rem; margin: 8px 0; }
  .pseudo-compare-grid { display: grid; grid-template-columns: 1fr; gap: 8px; margin-top: 8px; }
  .table-scroll { width: 100%; overflow-x: auto; margin-top: 6px; }
  .test-table { width: 100%; border-collapse: collapse; font-family: sans-serif; font-size: 0.8rem; background: white; }
  .test-table th { background: #f1f5f9; color: #1e293b; padding: 5px; border: 1px solid #cbd5e1; text-align: left; }
  .test-table td { padding: 5px; border: 1px solid #cbd5e1; font-family: monospace; }
  .stamp-banner { background: #dcfce7; border: 2px dashed #16a34a; color: #14532d; padding: 10px; border-radius: 8px; margin: 6px 0; font-family: sans-serif; font-size: 0.88rem; font-weight: bold; text-align: center; }
  .flow-oval { border: 2px solid #1e40af; background: #dbeafe; border-radius: 20px; padding: 4px 14px; display: inline-block; font-weight: bold; font-family: sans-serif; font-size: 0.82rem; }
  .flow-io { border: 2px solid #1e40af; background: #eff6ff; transform: skew(-10deg); padding: 4px 14px; display: inline-block; font-family: monospace; font-size: 0.82rem; }
  .flow-proc { border: 2px solid #1e40af; background: #ffffff; padding: 4px 14px; display: inline-block; font-family: monospace; font-size: 0.82rem; }
  .flow-arrow { font-size: 1rem; color: #1e40af; font-weight: bold; margin: 1px 0; }
  .lab-footer { text-align: center; color: #64748b; font-family: sans-serif; font-size: 0.78rem; margin-top: 14px; padding-top: 8px; border-top: 1px solid #cbd5e1; }
</style>
<div class='lab-header'>
  <div class='book-badge'>Algorithmic Thinking with Python</div>
  <h1>Chapter 1: The Recipe for Problem Solving</h1>
  <div class='author-line'>Dr. Ajith K K, Dept. of ECE, GCE Kannur</div>
</div>
""")

footer_widget = widgets.HTML(value="""
<div class='lab-footer'>
  <strong>Algorithmic Thinking with Python</strong> &bull; Dr. Ajith K K, Dept. of ECE, GCE Kannur
</div>
""")

roll_input = widgets.Text(
    placeholder="Roll No (e.g. 26EC012)",
    description="Roll No:",
    layout=widgets.Layout(width='250px', margin='0 0 6px 0')
)

# Native VBox with 4 individual 1-line HTML widgets -> Physically pushes Tabs down below it!
p_row1 = widgets.HTML(layout=widgets.Layout(margin='0'))
p_row2 = widgets.HTML(layout=widgets.Layout(margin='0'))
p_row3 = widgets.HTML(layout=widgets.Layout(margin='0'))
p_row4 = widgets.HTML(layout=widgets.Layout(margin='0'))

progress_box = widgets.VBox(
    [p_row1, p_row2, p_row3, p_row4],
    layout=widgets.Layout(
        background='#e2e8f0',
        border='1px solid #cbd5e1',
        padding='6px 10px',
        margin='0 0 10px 0',
        width='98%'
    )
)

stamp_box = widgets.HTML(value="")

tabs = widgets.Tab(layout=widgets.Layout(width='99%', margin='4px 0 0 0'))
stage2_container = widgets.VBox()
stage3_container = widgets.VBox()
stage2_content = widgets.VBox()
stage3_content = widgets.VBox()

stage2_lock_msg = widgets.HTML(value=f"""
<div class='lock-banner'>
  &#128274; <strong>Stage 2 Locked!</strong> Complete all {NUM_MCQ} MCQs in Tab 1 first.
</div>
""")

stage3_lock_msg = widgets.HTML(value=f"""
<div class='lock-banner'>
  &#128274; <strong>Stage 3 Locked!</strong> Complete all {NUM_PSEUDO} Pseudocode tasks in Tab 2 first.
</div>
""")

def refresh_progress(auto_switch=False):
    m_done = sum(state["mcq_passed"])
    p_done = sum(state["pseudo_done"])
    c_done = sum(state["py_passed"])
    tc_done = sum(state["py_tc_scores"])
    max_tc = NUM_PY * 3

    total_score = m_done * 1 + p_done * 2 + tc_done
    max_score = NUM_MCQ * 1 + NUM_PSEUDO * 2 + max_tc

    s1_icon = "&#9989;" if m_done == NUM_MCQ else "&#9203;"
    s2_icon = "&#9989;" if p_done == NUM_PSEUDO else ("&#9203;" if m_done == NUM_MCQ else "&#128274;")
    s3_icon = "&#9989;" if c_done == NUM_PY else ("&#9203;" if p_done == NUM_PSEUDO else "&#128274;")

    # Each row is 1 single line inside a native VBox -> Zero clipping on mobile!
    p_row1.value = f"<div class='prog-line'>{s1_icon} Stage 1 (MCQs): {m_done} / {NUM_MCQ}</div>"
    p_row2.value = f"<div class='prog-line'>{s2_icon} Stage 2 (Pseudocode): {p_done} / {NUM_PSEUDO}</div>"
    p_row3.value = f"<div class='prog-line'>{s3_icon} Stage 3 (Python): {c_done} / {NUM_PY} ({tc_done}/{max_tc} Tests)</div>"
    p_row4.value = f"<div class='prog-line-score'>&#127942; Total Score: {total_score} / {max_score}</div>"

    # Minimal, narrow tab titles
    tabs.set_title(0, f"1. MCQs ({m_done}/{NUM_MCQ})")

    if m_done == NUM_MCQ:
        stage2_container.children = [stage2_content]
        tabs.set_title(1, f"2. Pseudo ({p_done}/{NUM_PSEUDO})")
        if auto_switch and p_done == 0 and tabs.selected_index == 0:
            tabs.selected_index = 1
    else:
        stage2_container.children = [stage2_lock_msg]
        tabs.set_title(1, "🔒 2. Pseudo")

    if m_done == NUM_MCQ and p_done == NUM_PSEUDO:
        stage3_container.children = [stage3_content]
        tabs.set_title(2, f"3. Python ({c_done}/{NUM_PY})")
        if auto_switch and c_done == 0 and tabs.selected_index == 1:
            tabs.selected_index = 2
    else:
        stage3_container.children = [stage3_lock_msg]
        tabs.set_title(2, "🔒 3. Python")

    if c_done == NUM_PY:
        r = roll_input.value.strip().upper() or "STUDENT"
        code_num = sum(ord(ch) * (i + 3) for i, ch in enumerate(r)) % 9000 + 1000
        suffix = r[-3:] if len(r) >= 3 else r
        stamp_box.value = f"""
        <div class='stamp-banner'>
          &#127881; CHAPTER 1 MASTERED! Write this Verification Stamp in your book:<br> <span style='color:#1e40af;'>CH1-{suffix}-{code_num}</span> (Score: {total_score}/{max_score})
        </div>
        """

roll_input.observe(lambda _: refresh_progress(False), names='value')

# ==========================================================
# BUILD UI CARDS FROM QUESTION BANKS
# ==========================================================
mcq_cards = []
for idx, q in enumerate(mcq_data):
    h = f"<h4 style='margin:0 0 4px 0; font-family:sans-serif;'>{q['title']}</h4><p style='margin:0 0 6px 0; font-family:sans-serif; font-size:0.9rem;'>{q['prompt']}</p>"
    if q['code']:
        h += f"<div class='code-card'>{q['code']}</div>"
    rb = widgets.RadioButtons(options=q['options'], value=None, layout=widgets.Layout(width='98%'))
    btn = widgets.Button(description="Check", button_style="primary", layout=widgets.Layout(width='110px', height='32px'))
    fb_html = widgets.HTML(value="")

    def make_mcq_cb(i=idx, r=rb, fb=fb_html, item=q):
        def _cb(b):
            ans = b64_dec(item['b64_correct'])
            if r.value is None:
                fb.value = "<div class='fail-box'>Select an option first!</div>"
            elif r.value == ans:
                state["mcq_passed"][i] = True
                fb.value = f"<div class='pass-box'>&#9989; {item['feedback'][r.value]}</div>"
                refresh_progress(auto_switch=True)
            else:
                fb.value = f"<div class='fail-box'>&#10060; {item['feedback'][r.value]}</div>"
        return _cb

    btn.on_click(make_mcq_cb())
    mcq_cards.append(widgets.VBox([widgets.HTML(value=h), rb, btn, fb_html],
                     layout=widgets.Layout(border='1px solid #cbd5e1', padding='10px', margin='0 0 10px 0')))

stage1_container = widgets.VBox(mcq_cards)

def draw_flow(steps):
    nodes = []
    for kind, txt in steps:
        cls = 'flow-oval' if kind == 'oval' else ('flow-io' if kind == 'io' else 'flow-proc')
        nodes.append(f"<div><span class='{cls}'>{txt}</span></div>")
    return "<div style='text-align:center; padding:8px; background:#f8fafc; border:1px solid #cbd5e1; border-radius:6px;'>" + "<div class='flow-arrow'>&#8595;</div>".join(nodes) + "</div>"

pseudo_cards = []
for idx, p in enumerate(pseudo_tasks):
    ta = widgets.Textarea(placeholder="START\nREAD ...\nCOMPUTE ...\nPRINT ...\nSTOP", layout=widgets.Layout(width='98%', height='95px'))
    btn = widgets.Button(description="Compare Flowchart", button_style="primary", layout=widgets.Layout(width='160px', height='32px'))
    fb_html = widgets.HTML(value="")

    def make_p_cb(i=idx, t=ta, fb=fb_html, item=p):
        def _cb(b):
            val = t.value.strip()
            if len(val) < 15:
                fb.value = "<div class='fail-box'>Write your complete pseudocode steps first!</div>"
                return
            state["pseudo_done"][i] = True
            model_txt = b64_dec(item['b64_model'])
            f_html = draw_flow(item['flow'])
            fb.value = f"""
            <div class='pass-box'>&#9989; Submitted! Compare below:</div>
            <div class='pseudo-compare-grid'>
              <div><strong style='font-family:sans-serif; font-size:0.82rem;'>Your Pseudocode:</strong><div class='code-card'>{val}</div></div>
              <div><strong style='font-family:sans-serif; font-size:0.82rem;'>Instructor Model:</strong><div class='code-card'>{model_txt}</div></div>
            </div>
            <strong style='font-family:sans-serif; font-size:0.82rem;'>Flowchart:</strong>{f_html}
            """
            refresh_progress(auto_switch=True)
        return _cb

    btn.on_click(make_p_cb())
    pseudo_cards.append(widgets.VBox([widgets.HTML(value=f"<h4 style='margin:0 0 4px 0; font-family:sans-serif;'>{p['title']}</h4><p style='margin:0 0 6px 0; font-family:sans-serif; font-size:0.9rem;'>{p['prompt']}</p>"), ta, btn, fb_html],
                        layout=widgets.Layout(border='1px solid #cbd5e1', padding='10px', margin='0 0 10px 0')))

stage2_content.children = pseudo_cards

py_cards = []
for idx, (title, prompt, starter, grader, hint_txt) in enumerate(missions_spec):
    ta = widgets.Textarea(value=starter, layout=widgets.Layout(width='98%', height='105px'))
    btn = widgets.Button(description="Run Test Cases", button_style="success", layout=widgets.Layout(width='150px', height='32px'))
    fb = widgets.HTML(value="")

    def make_cb(i=idx, t=ta, f=fb, g=grader, h_str=hint_txt):
        def _cb(_):
            state["py_attempts"][i] += 1
            ok, tc_passed, table_html = g(t.value)
            state["py_passed"][i] = ok
            state["py_tc_scores"][i] = tc_passed
            banner = f"<div class='pass-box'>&#9989; PASSED ({tc_passed}/3)!</div>" if ok else f"<div class='fail-box'>&#10060; {tc_passed}/3 Passed:</div>"
            hint_html = ""
            if not ok and state["py_attempts"][i] >= 2:
                hint_html = f"<div class='hint-box'>&#128161; <strong>Hint:</strong> {h_str}</div>"
            f.value = banner + table_html + hint_html
            refresh_progress(auto_switch=False)
        return _cb

    btn.on_click(make_cb())
    py_cards.append(widgets.VBox([widgets.HTML(value=f"<h4 style='margin:0 0 4px 0; font-family:sans-serif;'>{title}</h4><p style='margin:0 0 6px 0; font-family:sans-serif; font-size:0.9rem;'>{prompt}</p>"), ta, btn, fb],
                    layout=widgets.Layout(border='1px solid #cbd5e1', padding='10px', margin='0 0 10px 0')))

stage3_content.children = py_cards

# ==========================================================
# INITIALIZE TABS & RENDER
# ==========================================================
tabs.children = [stage1_container, stage2_container, stage3_container]
refresh_progress(auto_switch=False)

display(widgets.VBox([header_widget, roll_input, progress_box, stamp_box, tabs, footer_widget]))
```

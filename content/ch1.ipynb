```python
import ipywidgets as widgets
from IPython.display import display
import base64
import math

# --- STYLING & HEADER ---
header_widget = widgets.HTML(value="""
<style>
  .lab-header { background: #1e40af; color: white; padding: 18px 22px; border-radius: 8px; margin-bottom: 18px; font-family: sans-serif; }
  .lab-header h1 { margin: 0 0 6px 0; font-size: 1.5rem; color: white; }
  .lab-header p { margin: 0; opacity: 0.9; font-size: 0.95rem; }
  .code-card { background: #1e293b; color: #f8fafc; padding: 12px; border-radius: 6px; font-family: monospace; white-space: pre-wrap; margin: 6px 0; line-height: 1.4; }
  .pass-box { background: #dcfce7; color: #166534; padding: 12px; border-radius: 6px; font-weight: 600; margin-top: 8px; border: 1px solid #86efac; font-family: sans-serif; }
  .fail-box { background: #fee2e2; color: #991b1b; padding: 12px; border-radius: 6px; font-weight: 600; margin-top: 8px; border: 1px solid #fca5a5; font-family: sans-serif; }
  .flow-oval { border: 2px solid #1e40af; background: #dbeafe; border-radius: 25px; padding: 6px 20px; display: inline-block; font-weight: bold; font-family: sans-serif; }
  .flow-io { border: 2px solid #1e40af; background: #eff6ff; transform: skew(-12deg); padding: 6px 20px; display: inline-block; font-family: monospace; }
  .flow-proc { border: 2px solid #1e40af; background: #ffffff; padding: 6px 20px; display: inline-block; font-family: monospace; }
  .flow-arrow { font-size: 1.2rem; color: #1e40af; font-weight: bold; margin: 2px 0; }
</style>
<div class='lab-header'>
  <h1>Chapter 1: The Recipe for Problem Solving</h1>
  <p>Switch between Part 1 (MCQs), Part 2 (Pseudocode & Flowcharts), and Part 3 (Python Autograder) below.</p>
</div>
""")

# --- PART 1: MCQs ---
mcq_data = [
    {
        "title": "1. Problem-Solving Strategies",
        "prompt": "You lose your hostel keys. Instead of searching every inch of campus, you check the canteen table and your pocket first. Which strategy is this?",
        "code": "",
        "options": ["Trial and Error", "Heuristics (Rule of Thumb)", "Working Backward", "Problem Decomposition"],
        "correct": "Heuristics (Rule of Thumb)",
        "feedback": {
            "Trial and Error": "Not quite! Trial and Error checks every possible spot without an educated guess.",
            "Heuristics (Rule of Thumb)": "Correct! Using past experience to check the most likely places first is a Heuristic.",
            "Working Backward": "Not quite! Working backward starts from the end goal (like an exam time) and subtracts.",
            "Problem Decomposition": "Not quite! Decomposition breaks a massive problem into smaller sub-problems."
        }
    },
    {
        "title": "2. The Text-to-Math Trap",
        "prompt": "What gets printed when Python runs these two lines?",
        "code": 'price = "45"\nprint(price + "5")',
        "options": ["50", "455", "price + 5", "TypeError"],
        "correct": "455",
        "feedback": {
            "50": "Careful! Look at the quotes around '45' and '5'. Python treats them as Strings (text), not numbers.",
            "455": "Spot on! Because both are Strings inside quotes, the + operator glues them side-by-side into '455'.",
            "price + 5": "price is not inside quotes in print(), so Python looks inside the price box.",
            "TypeError": "Gluing two Strings together with + is completely legal in Python."
        }
    },
    {
        "title": "3. Floor Division vs. Modulo",
        "prompt": "You divide 17 samosas equally among 5 friends. Which expression gives the LEFT OVER samosas?",
        "code": "",
        "options": ["17 / 5", "17 // 5", "17 % 5", "17 ** 5"],
        "correct": "17 % 5",
        "feedback": {
            "17 / 5": "17 / 5 gives 3.4 (decimal division).",
            "17 // 5": "Floor division (//) gives 3 (how many each friend eats, not what is left over).",
            "17 % 5": "Correct! Modulo (%) gives the remainder of division: 17 % 5 = 2 leftover samosas.",
            "17 ** 5": "** calculates 17 to the power of 5!"
        }
    },
    {
        "title": "4. Operator Precedence (BODMAS)",
        "prompt": "Act as the computer. What exact number is stored in result?",
        "code": "result = 10 - 2 * 3 ** 2",
        "options": ["64", "72", "-8", "-26"],
        "correct": "-8",
        "feedback": {
            "64": "Remember BODMAS! Do not subtract 10 - 2 first.",
            "72": "Python calculates 3 ** 2 (9) first, then 2 * 9 (18), then 10 - 18.",
            "-8": "Correct! Step 1: 3 ** 2 = 9. Step 2: 2 * 9 = 18. Step 3: 10 - 18 = -8.",
            "-26": "2 * 3 ** 2 squares only the 3, not (2 * 3)."
        }
    }
]

mcq_cards = []
for q in mcq_data:
    h = f"<h3 style='margin:0 0 6px 0;'>{q['title']}</h3><p style='margin:0 0 8px 0;'>{q['prompt']}</p>"
    if q['code']:
        h += f"<div class='code-card'>{q['code']}</div>"
    rb = widgets.RadioButtons(options=q['options'], value=None, layout=widgets.Layout(width='100%'))
    btn = widgets.Button(description="Check Answer", button_style="primary")
    fb_html = widgets.HTML(value="")

    def make_mcq_cb(r=rb, fb=fb_html, item=q):
        def _cb(b):
            if r.value is None:
                fb.value = "<div class='fail-box'>Please select an answer choice first!</div>"
            elif r.value == item['correct']:
                fb.value = f"<div class='pass-box'>&#9989; PASS: {item['feedback'][r.value]}</div>"
            else:
                fb.value = f"<div class='fail-box'>&#10060; TRY AGAIN: {item['feedback'][r.value]}</div>"
        return _cb

    btn.on_click(make_mcq_cb())
    mcq_cards.append(widgets.VBox([widgets.HTML(value=h), rb, btn, fb_html],
                     layout=widgets.Layout(border='1px solid #cbd5e1', padding='14px', margin='0 0 14px 0')))

tab1 = widgets.VBox(mcq_cards)

# --- PART 2: PSEUDOCODE & FLOWCHARTS ---
def draw_flow(steps):
    nodes = []
    for kind, txt in steps:
        cls = 'flow-oval' if kind == 'oval' else ('flow-io' if kind == 'io' else 'flow-proc')
        nodes.append(f"<div><span class='{cls}'>{txt}</span></div>")
    return "<div style='text-align:center; padding:12px; background:#f8fafc; border:1px solid #cbd5e1; border-radius:8px;'>" + "<div class='flow-arrow'>&#8595;</div>".join(nodes) + "</div>"

pseudo_tasks = [
    {
        "title": "1. Area of a Circle (AI Checkpoint 1.1)",
        "prompt": "Write a 4-step pseudocode algorithm to READ radius, COMPUTE area = 3.14159 * radius * radius, and PRINT area.",
        "b64_model": "U1RBUlQKUkVBRCBSYWRpdXMKQ09NUFVURSBhcmVhID0gMy4xNDE1OSAqIFJhZGl1cyAqIFJhZGl1cwpQUklOVCBhcmVhClNUT1A=",
        "flow": [("oval", "START"), ("io", "READ radius"), ("proc", "COMPUTE area = 3.14159 * radius * radius"), ("io", "PRINT area"), ("oval", "STOP")]
    },
    {
        "title": "2. Weeks & Leftover Days (Level 3 Mission)",
        "prompt": "Write pseudocode to READ total_days, compute full weeks using // 7, compute leftover days using % 7, and PRINT both.",
        "b64_model": "U1RBUlQKUkVBRCB0b3RhbF9kYXlzCkNPTVBVVEUgd2Vla3MgPSB0b3RhbF9kYXlzIC8vIDcKQ09NUFVURSBsZWZ0b3ZlciA9IHRvdGFsX2RheXMgJSA3ClBSSU5UIHdlZWtzLCBsZWZ0b3ZlcgpTVE9Q",
        "flow": [("oval", "START"), ("io", "READ total_days"), ("proc", "COMPUTE weeks = total_days // 7"), ("proc", "COMPUTE leftover = total_days % 7"), ("io", "PRINT weeks, leftover"), ("oval", "STOP")]
    }
]

pseudo_cards = []
for p in pseudo_tasks:
    ta = widgets.Textarea(placeholder="START\nREAD ...", layout=widgets.Layout(width='98%', height='110px'))
    btn = widgets.Button(description="Submit & Compare", button_style="primary")
    fb_html = widgets.HTML(value="")

    def make_p_cb(t=ta, fb=fb_html, item=p):
        def _cb(b):
            val = t.value.strip()
            if len(val) < 10:
                fb.value = "<div class='fail-box'>Please write out your complete pseudocode first!</div>"
                return
            model_txt = base64.b64decode(item['b64_model']).decode('utf-8')
            f_html = draw_flow(item['flow'])
            fb.value = f"""
            <div style='display:grid; grid-template-columns:1fr 1fr; gap:12px; margin-top:10px;'>
              <div><h4 style='margin:0;'>Your Pseudocode:</h4><div class='code-card'>{val}</div></div>
              <div><h4 style='margin:0;'>Instructor Model:</h4><div class='code-card'>{model_txt}</div></div>
            </div>
            <h4 style='margin:12px 0 4px 0;'>Corresponding Flowchart:</h4>{f_html}
            """
        return _cb

    btn.on_click(make_p_cb())
    pseudo_cards.append(widgets.VBox([widgets.HTML(value=f"<h3>{p['title']}</h3><p>{p['prompt']}</p>"), ta, btn, fb_html],
                        layout=widgets.Layout(border='1px solid #cbd5e1', padding='14px', margin='0 0 14px 0')))

tab2 = widgets.VBox(pseudo_cards)

# --- PART 3: PYTHON AUTOGRADER ---
def grade_m1(code):
    for tin, exp in [("100", 105), ("45", 50)]:
        ns = {"input": lambda p="": tin}
        out = []
        ns["print"] = lambda *a: out.append(" ".join(str(x) for x in a))
        try:
            exec(code, ns)
        except TypeError:
            return False, "TypeError! Did you wrap input() inside int()?"
        except Exception as e:
            return False, f"Error: {e}"
        if ns.get("final_bill") != exp:
            return False, f"For input {tin}, expected {exp}, got {ns.get('final_bill')}."
    return True, "Awesome! You melted the String into an Integer with int()!"

def grade_m2(code):
    for tin, ew, el in [("10", 1, 3), ("28", 4, 0)]:
        ns = {"input": lambda p="": tin}
        ns["print"] = lambda *a: None
        try:
            exec(code, ns)
        except Exception as e:
            return False, f"Error: {e}"
        if ns.get("weeks") != ew or ns.get("leftover") != el:
            return False, f"For {tin} days, expected weeks={ew} and leftover={el}."
    return True, "Brilliant! Floor Division (//) and Modulo (%) passed all tests!"

def grade_m3(code):
    if "import math" not in code:
        return False, "Add 'import math' at the top of your code!"
    if "math.pi" not in code:
        return False, "Use 'math.pi' instead of typing 3.14 manually!"
    for tin, rv in [("5", 5.0), ("2.5", 2.5)]:
        ns = {"input": lambda p="": tin, "print": lambda *a: None}
        try:
            exec(code, ns)
        except Exception as e:
            return False, f"Error: {e}"
        exp = math.pi * (rv ** 2)
        if "area" not in ns or abs(ns["area"] - exp) > 0.001:
            return False, f"For radius {rv}, expected area ~{exp:.2f}, got {ns.get('area')}."
    return True, "Perfect! You imported math and computed exact circle areas!"

py_tasks = [
    {
        "title": "Mission 1: Fix the Text-to-Math Trap",
        "prompt": "Wrap input() inside int() on line 1 so the Rs. 5 tax is added properly.",
        "starter": 'price = input("Enter item price: ")\ntax = 5\nfinal_bill = price + tax\nprint(final_bill)',
        "grader": grade_m1
    },
    {
        "title": "Mission 2: The Week & Day Splitter",
        "prompt": "Compute full 7-day weeks using // and leftover days using %.",
        "starter": 'days = int(input("Enter days: "))\nweeks = \nleftover = \nprint(weeks, leftover)',
        "grader": grade_m2
    },
    {
        "title": "Mission 3: Exact Circle Area Calculator",
        "prompt": "Import math at the top, then compute area = math.pi * (radius ** 2).",
        "starter": '# Step 1: Import math\n\nradius = float(input("Enter radius: "))\narea = \nprint(area)',
        "grader": grade_m3
    }
]

py_cards = []
for m in py_tasks:
    ta = widgets.Textarea(value=m['starter'], layout=widgets.Layout(width='98%', height='120px'))
    btn = widgets.Button(description="Run & Grade My Code", button_style="success")
    fb_html = widgets.HTML(value="")

    def make_py_cb(t=ta, fb=fb_html, g=m['grader']):
        def _cb(b):
            ok, msg = g(t.value)
            cls = 'pass-box' if ok else 'fail-box'
            prefix = '&#9989; PASS:' if ok else '&#10060; FAIL:'
            fb.value = f"<div class='{cls}'>{prefix} {msg}</div>"
        return _cb

    btn.on_click(make_py_cb())
    py_cards.append(widgets.VBox([widgets.HTML(value=f"<h3>{m['title']}</h3><p>{m['prompt']}</p>"), ta, btn, fb_html],
                    layout=widgets.Layout(border='1px solid #cbd5e1', padding='14px', margin='0 0 14px 0')))

tab3 = widgets.VBox(py_cards)

tabs = widgets.Tab(children=[tab1, tab2, tab3])
tabs.set_title(0, "Part 1: MCQs")
tabs.set_title(1, "Part 2: Pseudocode")
tabs.set_title(2, "Part 3: Python Code")

display(widgets.VBox([header_widget, tabs]))
```

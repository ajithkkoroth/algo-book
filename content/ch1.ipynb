```python
# ======================================================================
# CELL 2: AUTOMATIC RENDER ENGINE (DO NOT EDIT BELOW THIS LINE)
# Automatically encodes answers in Base64 and builds the fast UI.
# ======================================================================
import json
import base64
from IPython.display import display, HTML

def _b64(text):
    return base64.b64encode(text.encode('utf-8')).decode('utf-8')

# Automatically convert plain-text answers & models into Base64
encoded_mcqs = []
for q in MCQ_BANK:
    q_copy = dict(q)
    q_copy["b64_correct"] = _b64(q["correct"])
    del q_copy["correct"]
    encoded_mcqs.append(q_copy)

encoded_pseudos = []
for p in PSEUDO_BANK:
    p_copy = dict(p)
    p_copy["b64_model"] = _b64(p["model"])
    del p_copy["model"]
    encoded_pseudos.append(p_copy)

payload_json = json.dumps({
    "config": LAB_CONFIG,
    "mcqs": encoded_mcqs,
    "pseudos": encoded_pseudos,
    "missions": PYTHON_MISSIONS
})

html_template = r"""
<script src="https://cdn.jsdelivr.net/pyodide/v0.25.0/full/pyodide.js" async></script>
<style>
  .lab-wrap { max-width: 820px; margin: 0 auto; font-family: system-ui, -apple-system, sans-serif; color: #1e293b; }
  .lab-header { background: linear-gradient(135deg, #1e3a8a, #1e40af); color: white; padding: 14px 18px; border-radius: 8px; margin-bottom: 10px; }
  .book-badge { display: inline-block; background: rgba(255,255,255,0.18); color: #e0f2fe; padding: 2px 8px; border-radius: 12px; font-size: 0.72rem; font-weight: 700; text-transform: uppercase; margin-bottom: 4px; }
  .lab-header h1 { margin: 0 0 2px 0; font-size: 1.3rem; color: white; line-height: 1.25; }
  .author-line { margin: 0; font-size: 0.85rem; color: #bfdbfe; font-weight: 600; }

  .roll-bar { display: flex; align-items: center; gap: 8px; margin-bottom: 8px; flex-wrap: wrap; }
  .roll-bar input { padding: 6px 10px; border: 1px solid #cbd5e1; border-radius: 6px; font-size: 0.9rem; width: 180px; }
  .roll-bar input:disabled { background: #e2e8f0; color: #475569; font-weight: bold; }
  .btn-primary { background: #1e40af; color: white; border: none; padding: 7px 14px; border-radius: 6px; font-size: 0.85rem; font-weight: 600; cursor: pointer; }
  .btn-primary:disabled { background: #64748b; cursor: not-allowed; }
  .btn-success { background: #15803d; color: white; border: none; padding: 7px 14px; border-radius: 6px; font-size: 0.85rem; font-weight: 600; cursor: pointer; }

  .resp-progress-box {
    display: flex; flex-direction: row; justify-content: space-between; align-items: center;
    background: #e2e8f0; border: 1px solid #cbd5e1; border-radius: 8px;
    padding: 8px 14px; margin-bottom: 10px; font-size: 0.85rem; font-weight: 600; gap: 6px;
  }
  .prog-score { color: #1e40af; font-weight: 700; }

  .min-tabs { display: flex; gap: 4px; border-bottom: 2px solid #cbd5e1; margin-bottom: 12px; }
  .min-tab {
    background: #f1f5f9; color: #475569; border: 1px solid #cbd5e1; border-bottom: none;
    padding: 6px 10px; border-radius: 6px 6px 0 0; font-size: 0.8rem; font-weight: 700; cursor: pointer;
  }
  .min-tab.active { background: #1e40af; color: white; border-color: #1e40af; }

  .q-card { border: 1px solid #cbd5e1; border-radius: 8px; padding: 12px; margin-bottom: 12px; background: white; }
  .q-card h4 { margin: 0 0 4px 0; font-size: 0.98rem; }
  .q-card p { margin: 0 0 8px 0; font-size: 0.9rem; }
  .code-card { background: #1e293b; color: #f8fafc; padding: 10px; border-radius: 6px; font-family: monospace; font-size: 0.85rem; white-space: pre-wrap; word-break: break-word; margin: 6px 0; }
  .opt-label { display: block; padding: 6px 8px; margin: 4px 0; border: 1px solid #e2e8f0; border-radius: 6px; cursor: pointer; font-size: 0.88rem; }
  .opt-label:hover { background: #f8fafc; }
  textarea.code-in { width: 100%; box-sizing: border-box; font-family: monospace; font-size: 0.86rem; padding: 8px; border: 1px solid #cbd5e1; border-radius: 6px; background: #f8fafc; margin-bottom: 6px; }

  .pass-box { background: #dcfce7; color: #166534; padding: 8px 10px; border-radius: 6px; font-weight: 600; font-size: 0.85rem; margin-top: 6px; border: 1px solid #86efac; }
  .fail-box { background: #fee2e2; color: #991b1b; padding: 8px 10px; border-radius: 6px; font-weight: 600; font-size: 0.85rem; margin-top: 6px; border: 1px solid #fca5a5; }
  .hint-box { background: #fef9c3; color: #854d0e; padding: 8px 10px; border-radius: 6px; font-weight: 500; font-size: 0.85rem; margin-top: 6px; border: 1px solid #fde047; }
  .lock-banner { background: #f1f5f9; color: #475569; padding: 16px 12px; border-radius: 8px; border: 2px dashed #94a3b8; text-align: center; font-size: 0.9rem; margin: 8px 0; }
  .roll-gate-banner { background: #eff6ff; color: #1e3a8a; padding: 14px 12px; border-radius: 8px; border: 2px solid #93c5fd; text-align: center; font-size: 0.9rem; margin: 6px 0; }
  .stamp-banner { background: #dcfce7; border: 2px dashed #16a34a; color: #14532d; padding: 10px; border-radius: 8px; margin: 6px 0; font-size: 0.88rem; font-weight: bold; text-align: center; }

  .pseudo-compare-grid { display: grid; grid-template-columns: 1fr 1fr; gap: 12px; margin-top: 8px; }
  .table-scroll { width: 100%; overflow-x: auto; margin-top: 6px; }
  .test-table { width: 100%; border-collapse: collapse; font-size: 0.82rem; background: white; }
  .test-table th { background: #f1f5f9; color: #1e293b; padding: 6px; border: 1px solid #cbd5e1; text-align: left; }
  .test-table td { padding: 6px; border: 1px solid #cbd5e1; font-family: monospace; }
  .flow-oval { border: 2px solid #1e40af; background: #dbeafe; border-radius: 20px; padding: 4px 14px; display: inline-block; font-weight: bold; font-size: 0.82rem; }
  .flow-io { border: 2px solid #1e40af; background: #eff6ff; transform: skew(-10deg); padding: 4px 14px; display: inline-block; font-family: monospace; font-size: 0.82rem; }
  .flow-proc { border: 2px solid #1e40af; background: #ffffff; padding: 4px 14px; display: inline-block; font-family: monospace; font-size: 0.82rem; }
  .flow-arrow { font-size: 1rem; color: #1e40af; font-weight: bold; margin: 1px 0; }
  .lab-footer { text-align: center; color: #64748b; font-size: 0.78rem; margin-top: 14px; padding-top: 8px; border-top: 1px solid #cbd5e1; }

  @media (max-width: 640px) {
    .resp-progress-box { flex-direction: column; align-items: flex-start; padding: 6px 10px; gap: 3px; }
    .pseudo-compare-grid { grid-template-columns: 1fr; }
    .lab-header h1 { font-size: 1.15rem; }
  }
</style>

<div class='lab-wrap'>
  <div class='lab-header'>
    <div id='ui-book-badge' class='book-badge'></div>
    <h1 id='ui-chap-title'></h1>
    <div id='ui-author-line' class='author-line'></div>
  </div>

  <div class='roll-bar'>
    <label id='ui-roll-lbl' style='font-size:0.88rem; font-weight:600;'></label>
    <input type='text' id='roll-in'>
    <button id='roll-btn' class='btn-primary' onclick='lockRoll()'></button>
  </div>
  <div id='roll-err'></div>
  <div id='prog-box' class='resp-progress-box'></div>
  <div id='stamp-box'></div>
  <div id='roll-gate' class='roll-gate-banner'></div>

  <div id='workspace' style='display:none;'>
    <div class='min-tabs'>
      <button id='tab-btn-0' class='min-tab active' onclick='switchTab(0)'></button>
      <button id='tab-btn-1' class='min-tab' onclick='switchTab(1)'></button>
      <button id='tab-btn-2' class='min-tab' onclick='switchTab(2)'></button>
    </div>
    <div id='tab-pane-0'></div>
    <div id='tab-pane-1' style='display:none;'></div>
    <div id='tab-pane-2' style='display:none;'></div>
  </div>
  <div id='ui-footer' class='lab-footer'></div>
</div>

<script>
const DATA = __PAYLOAD_JSON__;
const CFG = DATA.config;
const mcqData = DATA.mcqs;
const pseudoTasks = DATA.pseudos;
const pyMissions = DATA.missions;

// Populate Configurable Texts
document.getElementById('ui-book-badge').innerHTML = CFG.book_title;
document.getElementById('ui-chap-title').innerHTML = CFG.chapter_title;
document.getElementById('ui-author-line').innerHTML = CFG.author_line;
document.getElementById('ui-roll-lbl').innerHTML = CFG.roll_label;
document.getElementById('roll-in').placeholder = CFG.roll_placeholder;
document.getElementById('roll-btn').innerHTML = CFG.lock_btn_text;
document.getElementById('roll-gate').innerHTML = CFG.roll_gate_msg;
document.getElementById('ui-footer').innerHTML = CFG.footer_text;

const NUM_MCQ = mcqData.length;
const NUM_PSEUDO = pseudoTasks.length;
const NUM_PY = pyMissions.length;

const state = {
  rollLocked: false,
  lockedRollStr: "",
  activeTab: 0,
  mcqPassed: Array(NUM_MCQ).fill(false),
  pseudoDone: Array(NUM_PSEUDO).fill(false),
  pyPassed: Array(NUM_PY).fill(false),
  pyTcScores: Array(NUM_PY).fill(0),
  pyAttempts: Array(NUM_PY).fill(0)
};

function lockRoll() {
  const rIn = document.getElementById('roll-in');
  const val = rIn.value.trim().toUpperCase();
  const err = document.getElementById('roll-err');
  if (val.length < 1) {
    err.innerHTML = "<div class='fail-box' style='margin-bottom:8px;'>Please enter your Roll Number before locking!</div>";
    return;
  }
  err.innerHTML = "";
  state.rollLocked = true;
  state.lockedRollStr = val;
  rIn.value = val;
  rIn.disabled = true;
  const btn = document.getElementById('roll-btn');
  btn.disabled = true;
  btn.innerHTML = CFG.locked_btn_text;
  document.getElementById('roll-gate').style.display = 'none';
  document.getElementById('workspace').style.display = 'block';
  refreshProgress(false);
}

function switchTab(idx) {
  state.activeTab = idx;
  for (let i = 0; i < 3; i++) {
    document.getElementById(`tab-btn-${i}`).classList.toggle('active', i === idx);
    document.getElementById(`tab-pane-${i}`).style.display = (i === idx) ? 'block' : 'none';
  }
}

function refreshProgress(autoSwitch) {
  const mDone = state.mcqPassed.filter(Boolean).length;
  const pDone = state.pseudoDone.filter(Boolean).length;
  const cDone = state.pyPassed.filter(Boolean).length;
  const tcDone = state.pyTcScores.reduce((a, b) => a + b, 0);
  const maxTc = pyMissions.reduce((acc, m) => acc + m.tests.length, 0);
  const totalScore = mDone * 1 + pDone * 2 + tcDone;
  const maxScore = NUM_MCQ * 1 + NUM_PSEUDO * 2 + maxTc;

  const s1Icon = (mDone === NUM_MCQ) ? "&#9989;" : "&#9203;";
  const s2Icon = (pDone === NUM_PSEUDO) ? "&#9989;" : ((mDone === NUM_MCQ) ? "&#9203;" : "&#128274;");
  const s3Icon = (cDone === NUM_PY) ? "&#9989;" : ((pDone === NUM_PSEUDO) ? "&#9203;" : "&#128274;");

  document.getElementById('prog-box').innerHTML = `
    <div>${s1Icon} Stage 1 (MCQs): ${mDone}/${NUM_MCQ}</div>
    <div>${s2Icon} Stage 2 (Pseudocode): ${pDone}/${NUM_PSEUDO}</div>
    <div>${s3Icon} Stage 3 (Python): ${cDone}/${NUM_PY} (${tcDone}/${maxTc} Tests)</div>
    <div class='prog-score'>&#127942; Score: ${totalScore}/${maxScore}</div>
  `;

  document.getElementById('tab-btn-0').innerText = `1. MCQs (${mDone}/${NUM_MCQ})`;
  document.getElementById('tab-btn-1').innerText = (mDone === NUM_MCQ) ? `2. Pseudocode (${pDone}/${NUM_PSEUDO})` : `🔒 2. Pseudocode`;
  document.getElementById('tab-btn-2').innerText = (mDone === NUM_MCQ && pDone === NUM_PSEUDO) ? `3. Python (${cDone}/${NUM_PY})` : `🔒 3. Python`;

  document.getElementById('stage2-lock').style.display = (mDone === NUM_MCQ) ? 'none' : 'block';
  document.getElementById('stage2-cards').style.display = (mDone === NUM_MCQ) ? 'block' : 'none';

  const s3Unlocked = (mDone === NUM_MCQ && pDone === NUM_PSEUDO);
  document.getElementById('stage3-lock').style.display = s3Unlocked ? 'none' : 'block';
  document.getElementById('stage3-cards').style.display = s3Unlocked ? 'block' : 'none';

  if (autoSwitch && mDone === NUM_MCQ && pDone === 0 && state.activeTab === 0) switchTab(1);
  if (autoSwitch && s3Unlocked && cDone === 0 && state.activeTab === 1) switchTab(2);

  if (cDone === NUM_PY && state.rollLocked) {
    const r = state.lockedRollStr;
    let sum = 0;
    for (let i = 0; i < r.length; i++) sum += r.charCodeAt(i) * (i + 3);
    const codeNum = (sum % 9000) + 1000;
    const suffix = r.length >= 3 ? r.slice(-3) : r;
    document.getElementById('stamp-box').innerHTML = `
      <div class='stamp-banner'>
        &#127881; ${CFG.chapter_code} MASTERED!: <span style='color:#1e40af;'>${CFG.chapter_code}-${suffix}-${codeNum}</span> 
      </div>`;
  }
}

// Render MCQs
const p0 = document.getElementById('tab-pane-0');
mcqData.forEach((q, idx) => {
  const codeHtml = q.code ? `<div class='code-card'>${q.code}</div>` : "";
  const optsHtml = q.options.map(opt => `
    <label class='opt-label'>
      <input type='radio' name='mcq-${idx}' value="${opt.replace(/"/g, '&quot;')}"> ${opt}
    </label>`).join("");
  p0.innerHTML += `
    <div class='q-card'>
      <h4>${q.title}</h4>
      <p>${q.prompt}</p>
      ${codeHtml}
      ${optsHtml}
      <button class='btn-primary' onclick='checkMcq(${idx})'>Check</button>
      <div id='mcq-fb-${idx}'></div>
    </div>`;
});

function checkMcq(idx) {
  const q = mcqData[idx];
  const sel = document.querySelector(`input[name="mcq-${idx}"]:checked`);
  const fb = document.getElementById(`mcq-fb-${idx}`);
  if (!sel) { fb.innerHTML = "<div class='fail-box'>Select an option first!</div>"; return; }
  const ans = atob(q.b64_correct);
  if (sel.value === ans) {
    state.mcqPassed[idx] = true;
    fb.innerHTML = `<div class='pass-box'>&#9989; ${q.feedback[sel.value]}</div>`;
    refreshProgress(true);
  } else {
    fb.innerHTML = `<div class='fail-box'>&#10060; ${q.feedback[sel.value]}</div>`;
  }
}

// Render Pseudocode
const p1 = document.getElementById('tab-pane-1');
p1.innerHTML = `<div id='stage2-lock' class='lock-banner'>${CFG.stage2_lock_msg}</div><div id='stage2-cards' style='display:none;'></div>`;
const s2Cards = document.getElementById('stage2-cards');
pseudoTasks.forEach((p, idx) => {
  s2Cards.innerHTML += `
    <div class='q-card'>
      <h4>${p.title}</h4>
      <p>${p.prompt}</p>
      <textarea id='ps-in-${idx}' class='code-in' rows='5' placeholder='START&#10;READ ...&#10;COMPUTE ...&#10;PRINT ...&#10;STOP'></textarea>
      <button class='btn-primary' onclick='checkPseudo(${idx})'>Compare Flowchart</button>
      <div id='ps-fb-${idx}'></div>
    </div>`;
});

function checkPseudo(idx) {
  const p = pseudoTasks[idx];
  const val = document.getElementById(`ps-in-${idx}`).value.trim();
  const fb = document.getElementById(`ps-fb-${idx}`);
  if (val.length < 15) { fb.innerHTML = "<div class='fail-box'>Write your complete pseudocode steps first!</div>"; return; }
  state.pseudoDone[idx] = true;
  const modelTxt = atob(p.b64_model);
  const nodes = p.flow.map(([k, t]) => {
    const cls = k === 'oval' ? 'flow-oval' : (k === 'io' ? 'flow-io' : 'flow-proc');
    return `<div><span class='${cls}'>${t}</span></div>`;
  }).join("<div class='flow-arrow'>&#8595;</div>");
  fb.innerHTML = `
    <div class='pass-box'>&#9989; Submitted! Compare below:</div>
    <div class='pseudo-compare-grid'>
      <div><strong style='font-size:0.82rem;'>Your Pseudocode:</strong><div class='code-card'>${val.replace(/</g,'&lt;')}</div></div>
      <div><strong style='font-size:0.82rem;'>Instructor Model:</strong><div class='code-card'>${modelTxt}</div></div>
    </div>
    <strong style='font-size:0.82rem;'>Flowchart:</strong>
    <div style='text-align:center; padding:8px; background:#f8fafc; border:1px solid #cbd5e1; border-radius:6px;'>${nodes}</div>`;
  refreshProgress(true);
}

// Render Python Missions
const p2 = document.getElementById('tab-pane-2');
p2.innerHTML = `<div id='stage3-lock' class='lock-banner'>${CFG.stage3_lock_msg}</div><div id='stage3-cards' style='display:none;'></div>`;
const s3Cards = document.getElementById('stage3-cards');
pyMissions.forEach((m, idx) => {
  s3Cards.innerHTML += `
    <div class='q-card'>
      <h4>${m.title}</h4>
      <p>${m.prompt}</p>
      <textarea id='py-in-${idx}' class='code-in' rows='5'>${m.starter}</textarea>
      <button class='btn-success' onclick='runPy(${idx})'>Run Test Cases</button>
      <div id='py-fb-${idx}'></div>
    </div>`;
});

let pyodideReady = null;
async function ensurePyodide() {
  if (pyodideReady) return pyodideReady;
  if (typeof loadPyodide !== 'undefined') {
    pyodideReady = await loadPyodide();
    return pyodideReady;
  }
  return null;
}
setTimeout(ensurePyodide, 800);

const GENERIC_GRADER = `
import json
spec = json.loads(_spec_json)
req = spec.get('required_code', [])
missing = [kw for kw in req if kw not in _student_code]
rows, passed_count = [], 0
for idx, tc in enumerate(spec['tests'], 1):
    tin = tc['input']
    exp_p = tc['expected_print']
    if missing:
        rows.append((idx, tin, exp_p, 'Missing keyword', False, spec.get('missing_code_msg', 'Check syntax')))
        continue
    ns = {'input': lambda p='': tin}
    out = []
    ns['print'] = lambda *a: out.append(' '.join(str(x) for x in a))
    try:
        exec(_student_code, ns)
        actual = out[-1].strip() if out else 'Nothing'
        vars_ok = True
        for vname, vexp in tc.get('check_vars', {}).items():
            vval = ns.get(vname)
            if spec.get('float_tolerance'):
                if not isinstance(vval, (int, float)) or abs(vval - vexp) > 0.001:
                    vars_ok = False
                else:
                    actual = f'{vval:.4f}'
            else:
                if isinstance(vval, float) and isinstance(vexp, int):
                    vars_ok = False
                elif vval != vexp:
                    vars_ok = False
        print_ok = True if spec.get('float_tolerance') else (actual == exp_p)
        passed = vars_ok and print_ok
        comment = tc['note'] if passed else spec.get('missing_code_msg', 'Check calculation & variables')
    except Exception as e:
        actual, passed, comment = type(e).__name__, False, str(e)
    if passed: passed_count += 1
    rows.append((idx, tin, exp_p, actual, passed, comment))
_res = (passed_count, rows)
`;

async function runPy(idx) {
  const m = pyMissions[idx];
  const code = document.getElementById(`py-in-${idx}`).value;
  const fb = document.getElementById(`py-fb-${idx}`);
  fb.innerHTML = "<div class='hint-box'>Running test cases...</div>";
  const py = await ensurePyodide();
  if (!py) { fb.innerHTML = "<div class='fail-box'>Python engine warming up—click again in 2 seconds!</div>"; return; }
  state.pyAttempts[idx] += 1;
  try {
    py.globals.set('_student_code', code);
    py.globals.set('_spec_json', JSON.stringify(m));
    await py.runPythonAsync(GENERIC_GRADER);
    const res = py.globals.get('_res').toJs();
    const tcPassed = res[0];
    const rows = res[1];
    const totalTc = m.tests.length;
    const ok = (tcPassed === totalTc);
    state.pyPassed[idx] = ok;
    state.pyTcScores[idx] = tcPassed;
    const trHtml = rows.map(r => {
      const badge = r[4] ? "<span style='color:#166534; font-weight:bold;'>&#9989; PASS</span>" : "<span style='color:#991b1b; font-weight:bold;'>&#10060; FAIL</span>";
      return `<tr><td>#${r[0]}</td><td><code>${r[1]}</code></td><td><code>${r[2]}</code></td><td><code>${r[3]}</code></td><td>${badge}</td><td style='font-family:sans-serif;'>${r[5]}</td></tr>`;
    }).join("");
    const banner = ok ? `<div class='pass-box'>&#9989; PASSED (${tcPassed}/${totalTc})!</div>` : `<div class='fail-box'>&#10060; ${tcPassed}/${totalTc} Passed:</div>`;
    const hintHtml = (!ok && state.pyAttempts[idx] >= 2) ? `<div class='hint-box'>&#128161; <strong>Hint:</strong> ${m.hint}</div>` : "";
    fb.innerHTML = `${banner}<div class='table-scroll'><table class='test-table'><thead><tr><th>#</th><th>Input</th><th>Expected</th><th>Yours</th><th>Status</th><th>Feedback</th></tr></thead><tbody>${trHtml}</tbody></table></div>${hintHtml}`;
    refreshProgress(false);
  } catch (e) {
    fb.innerHTML = `<div class='fail-box'>Error: ${e.message}</div>`;
  }
}

refreshProgress(false);
</script>
""".replace("__PAYLOAD_JSON__", payload_json)

display(HTML(html_template))
```
